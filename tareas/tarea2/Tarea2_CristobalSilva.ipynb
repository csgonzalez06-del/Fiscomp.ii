```python
L = [1.0, 2.5, 0.8, 5.0, 10.0] # Muestra de luminosidad en un grupo de estrellas
d = [10, 25, 15, 40, 60] # Muestra de distancias en parsecs de un grupo de estrellas
```

```python
def flujo_observado(L,d):
    """
    ---------
    Esta funcion calcula el flujo observado de las estrellas a partir de su luminosidad y distancia
    
    Se utilizan los siguientes parametros:
    ---------

    L : Lista de luminosidad de las estrellas 
    
    d: Lista de la distancia en parsecs de las estrellas 
    
    Retorna:
    La funcion retorna una lista con el flujo observado de las estrellas calculado mediante la formula: F = L / (4*pi*d^2)
    ----------

    """ 
    # Las listas de luminosidad y distancia (L y d) se transforman en arreglos de numpy 
    Luminosidad = np.array(L)
    distancia  = np.array(d)

    # Pasar de parsecs a metros
    distancia_metros = distancia * 3.086e16
    
    
    # Calcular el flujo observado, se utiliza la formula: F = L / (4*pi*d^2)
    Flujo_obs = Luminosidad / (4 * np.pi * distancia_metros**2)
    return Flujo_obs


# Se calcula el flujo observado de las estrellas y se imprime el resultado 
Lista_de_flujo = flujo_observado(L, d)
print("El flujo observado de las estrellas es:", Lista_de_flujo)


#Se calcula el flujo maximo, minimo, promedio y mediano mediante las funciones de numpy y se imprimen los resultados 
Maximo = np.max(flujo_observado(L, d))
Minimo = np.min(flujo_observado(L, d))
promedio = np.mean(flujo_observado(L, d))
mediana = np.median(flujo_observado(L, d))

print("El flujo maximo observado es:", Maximo)
print("El flujo minimo observado es:", Minimo)
print("El flujo promedio es:", promedio)
print("El flujo mediano es:", mediana)

```

```python
# Se muestra los indices de todos los elementos 
indices = np.arange(len(flujo_observado(L, d)))
print("Los indices de los elementos son:", indices)
```

```python
#Se crea una mascara para detectar los flujos menores a 1e-10
mascara_deteccion = np.array(flujo_observado(L, d) < 1e-10)
print("La mascara de deteccion es:", mascara_deteccion)
```

```python


# Se define la desviacion estandar  y la disminucion del flujo durante el transito
desviacion_estandar = 0.002 
disminucion_flujo = 0.02 

def simular_transito_estrella():
    """  
    Esta funcion simula el flujo de una estrella durante un periodo de 10 años.
    Se genara un arreglo de 500 elementos entre 0 y 10, el cual simula un intervalo de 10 años. Ademas se genera un arreglo de 500 elementos con valor 1, el cual simula el flujo de la estrella durante los 10 años.
    Se agrega un flujo de ruido gaussiano, el cual simulara el transito de la estrella, se simulara entre los 4.5 y 5 años, el cual tendra una desviacion estandar de 0.002. Se agregara un flujo de ruido gaussiano a la simulacion, el cual tendra una desviacion estandar de 0.002.
    -----------

    Parametros:
    -----------

    Desviacion estandar del ruido gaussiano: 0.002
    Tiempo de transito: entre 4.5 y 5 años
    Disminucion del flujo durante el transito: 0.02
    
    ----------
    Retorna: 
    La funcion retorna un arreglo de 500 elementos que simula el flujo de la estrella durante los 10 años, con un transito simulado entre los 4.5 y 5 años, con una disminucion del flujo de 0.02 y un flujo de ruido gaussiano con desviacion estandar de 0.002.

    """

    tiempo = np.linspace(0,10, 500)# se genera un arreglo de 500 elementos entre 0 y 10 que simula intervalo de diez años, compuesto por 500 mediciones equidistantes.
    flujo_estrella = np.ones_like(tiempo)# se genera un arreglo de 500 elementos con valor 1 que simula el flujo de la estrella durante los 10 años.

    # Se genera un flujo de ruido gaussiano con desviacion estandar de 0.002 y se agrega al flujo de la estrella
    ruido = np.random.normal(0, desviacion_estandar, len(tiempo)) 
    flujo = flujo_estrella + ruido 

    # Se simula el transito de la estrella entre los 4.5 y 5 años, disminuyendo el flujo en 0.02
    mascara_transito = (tiempo >= 4.5) & (tiempo <= 5)
    flujo[mascara_transito] -= disminucion_flujo

    return tiempo, flujo



```

```python
# se genera la mascara para el transito de la estrella, la cual se utilizara para el analisis de los resultados
tiempo, flujo = simular_transito_estrella()
mascara_transito = (tiempo >= 4.5) & (tiempo <= 5) 
```

```python

# Se calcula el flujo promedio, la desviacion estandar del flujo, el valor minimo del flujo y el tiempo en el que se produce el valor minimo del flujo y se inprimen los resultados
Flujo_promedio = np.mean(flujo) 
Desviacion_estandar_flujo = np.std(flujo)
valor_minimo = np.min(flujo) 
indice_minimo = np.argmin(flujo)
tiempo_minimo = tiempo[indice_minimo]


print("El flujo promedio es:", Flujo_promedio)
print("La desviacion estandar del flujo es:", Desviacion_estandar_flujo)
print("El valor minimo del flujo es:", valor_minimo)        
print("El tiempo en el que se produce el valor minimo del flujo es:", tiempo_minimo)
```

```python
# Se crea una mascara para detectar los flujos menos al parametro solicitado
mascara = flujo < (Flujo_promedio - 3 * desviacion_estandar)
print("La mascara de deteccion de flujos menores al promedio menos 3 veces la desviacion estandar es:", mascara)

# Se determina los puntos que son posibles transitos, es decir, los puntos que cumplen con la condicion de ser menores al promedio menos 3 veces la desviacion estandar y se imprimen los resultados
puntos_transito = tiempo[mascara]
print("Los puntos que son posibles transitos son:", puntos_transito)
```

```python
#Se estima la profundidad del transito mediante 
f_transito = np.mean(flujo[mascara_transito])
delta_f = 1 - f_transito
razon_radios = np.sqrt(delta_f)
print("La profundidad del transito es:", delta_f)   
print("La razon de los radios es:", razon_radios)
```

```python
# Compara los tiempos mediante el criterio 3 veces la desviacion estandar con el intervalo original de 4.5 a 5 años y se imprime el resultado
tiempo_inicial_det = puntos_transito.min()
tiempo_final_det = puntos_transito.max()
print(f" Detectado: {tiempo_inicial_det:.3f} a {tiempo_final_det:.3f} años (real: 4.5 a 5.0)")
print(" Puntos reales en el tránsito:", np.sum(mascara_transito))
print("Detectados dentro del intervalo real:", np.sum(mascara & mascara_transito))
print("Falsos positivos (fuera del tránsito):", np.sum(mascara & mascara_transito))
```
